```python
import random

class Jugador:
    def __init__(self, nombre):
        self.nombre = nombre
        self.monedas = 100
        self.simbolos = ["@", "#", "$", "%"]

    def apostar(self):
        self.monedas -= 10
        suerte = [random.choice(self.simbolos) for i in range(3)]
        return suerte

    def recibir_premio(self, lista):
        print(f"salieron: {lista}")
        if lista[0] == lista[1] == lista[2]:
            print("ez buenaaa 3 iguales! te llevas 50")
            self.monedas += 50
        elif lista[0] == lista[1] or lista[1] == lista[2] or lista[0] == lista[2]:
            print("algo ez algo 2 iguales! te llevas 20")
            self.monedas += 20
        else:
            print("no ganaste ni un centavo, brither")

    def mostrar_saldo(self):
        print(f"tienes ahorita: {self.monedas} monedas")
```

```python
def ver_si_puede_jugar(j):
    if j.monedas < 10:
        print("ya no tienes plata kbron")
        return False
    return True
```

```python
n = input("escribe tu nombre: ")
user = Jugador(n)

print(f"hola {user.nombre} empezamos con 100")

sigue = True
while sigue:
    if not ver_si_puede_jugar(user):
        break
        
    user.mostrar_saldo()
    quiere = input("quieres tirar? (s/n): ").lower()
    
    if quiere == "s":
        tiro = user.apostar()
        user.recibir_premio(tiro)
    elif quiere == "n":
        print(f"te vas con {user.monedas} monedas, chao")
        sigue = False
    else:
        print("esa letra no sirve pon s o n")

print("fin del juego")
```
