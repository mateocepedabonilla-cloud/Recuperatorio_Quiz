```python
import random

class Mazo:
    def __init__(self):
        self.baraja = [2,3,4,5,6,7,8,9,10,10,10,10,11] * 4
        random.shuffle(self.baraja)

    def entregar(self):
        return self.baraja.pop()

class Jugador:
    def __init__(self, nombre):
        self.nombre = nombre
        self.cartas = []
        self.puntos = 0

    def calcular_puntos(self):
        self.puntos = sum(self.cartas)
        if self.puntos > 21 and 11 in self.cartas:
            self.puntos -= 10
        return self.puntos

    def pedir_carta(self, mazo):
        nueva = mazo.entregar()
        self.cartas.append(nueva)
        self.calcular_puntos()

    def mostrar_mano(self):
        print(f"Tus cartas: {self.cartas} -> Suma: {self.puntos}")
        print(f"mano de {self.nombre}: {self.cartas} | puntos: {self.puntos}")
```

```python
def ver_si_perdio(p):
    if p.puntos > 21:
        p.mostrar_mano()
        print("perdiste porque te pasaste de 21")
        return True
    elif p.puntos == 21:
        p.mostrar_mano()
        print("sacaste 21, ganaste!!")
        return True
    return False
```

```python
nom = input("pon tu nombre: ")
usuario = Jugador(nom)
el_mazo = Mazo()

usuario.pedir_carta(el_mazo)
usuario.pedir_carta(el_mazo)

sigue = True
while sigue:
    usuario.mostrar_mano()
    if ver_si_perdio(usuario):
        break
    opcion = input("¿Pedir carta o plantarse? (p/s): ").lower()
    if opcion == 'p':
        usuario.pedir_carta(el_mazo)
        print("Nueva carta recibida...")
    elif opcion == 's':
        print(f"¡Te plantaste con {usuario.puntos}!")
        sigue = False
    else:
        print("esa tecla no es p ni s")

print("fin de la partida")
```
